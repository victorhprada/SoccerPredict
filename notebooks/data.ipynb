```python
import pandas as pd

# Vamos definir uma função simples para puxar dados
# Fonte: Repositório público de dados de futebol (exemplo com dados históricos)
# Usaremos dados reais hospedados no GitHub para facilitar o início

def carregar_dados_brasileirao():
    print("Iniciando download dos dados...")
    
    # Lista de URLs com dados do campeonato brasileiro (Série A) de 2020 a 2023
    # Estes são links diretos para arquivos CSV mantidos pela comunidade (football-data.co.uk ou similares)
    urls = [
        "https://www.football-data.co.uk/new/BRA.csv" 
    ]
    
    lista_dfs = []
    
    for url in urls:
        try:
            # O pandas lê o CSV direto da internet
            df = pd.read_csv(url)
            lista_dfs.append(df)
            print(f"Sucesso ao baixar: {url}")
        except Exception as e:
            print(f"Erro ao baixar {url}: {e}")
            
    if lista_dfs:
        # Junta tudo numa tabela só
        df_final = pd.concat(lista_dfs, ignore_index=True)
        
        # Filtra apenas o que parece ser Série A (ou limpa dados vazios)
        # O arquivo BRA.csv contém várias ligas, vamos olhar só as colunas principais
        return df_final
    else:
        return None

# Executa a função
dados = carregar_dados_brasileirao()

# Mostra as primeiras 5 linhas para você ver que funcionou
if dados is not None:
    print("\nDados carregados com sucesso! Aqui está uma amostra:")
    display(dados.head())
    print(f"\nTotal de partidas encontradas: {dados.shape[0]}")
else:
    print("Não foi possível carregar os dados.")
```

```python
# Vamos criar uma tabela simples de classificação de mandantes
tabela_mandante = dados.groupby('Home')['HG'].sum().sort_values(ascending=False)

print("Times com mais gols marcados em casa no histórico baixado:")
print(tabela_mandante.head(10))

# Vamos ver a média de gols por partida
media_gols = (dados['HG'] + dados['AG']).mean()
print(f"\n A média de gols por partida no Brasil é: {media_gols:.2f}")
```

```python
# 1. Selecionar apenas colunas essenciais
# Date = Data, Home = Mandante, Away = Visitante, HG = Gols Casa, AG = Gols Fora, Res = Resultado
colunas_desejadas = ['Date', 'Home', 'Away', 'HG', 'AG', 'Res']

# Criamos um novo dataframe (df_limpo) apenas com essas colunas
# O .copy() é importante para evitar avisos de alteração no original
df_limpo = dados[colunas_desejadas].copy()

# 2. Renomear para português para facilitar
df_limpo.columns = ['Data', 'Mandante', 'Visitante', 'Gols_Mandante', 'Gols_Visitante', 'Resultado']

# 3. Converter a coluna 'Data' que vem como texto para o formato de DATA real que o Python entende
# O formato mixed permite que o Pandas tente adivinhar se é dia/mês ou mês/dia
df_limpo['Data'] = pd.to_datetime(df_limpo['Data'], dayfirst=True)

# 4. Ordenar por data (do jogo mais antigo para o mais novo)
df_limpo = df_limpo.sort_values('Data')

# Mostra como ficou
print("Tabela limpa e traduzida:")
display(df_limpo.head())
print(df_limpo.tail()) # Mostra os últimos jogos para ver até que ano temos dados
```

```python
# Mapeamento: Dicionário que diz "Troque H por 1, D por 0, A por 2"
# Nota: A escolha dos números é arbitrária, mas precisa ser consistente.
mapa_resultado = {'H': 1, 'D': 0, 'A': 2}

# Aplica o mapeamento numa nova coluna chamada 'Alvo' (o que queremos prever)
df_limpo['Alvo'] = df_limpo['Resultado'].map(mapa_resultado)

# Vamos criar uma coluna de "Total de Gols" também, útil para previsões de Over/Under
df_limpo['Total_Gols'] = df_limpo['Gols_Mandante'] + df_limpo['Gols_Visitante']

display(df_limpo.head())
```

```python
# Função para calcular média acumulada de gols feitos até aquele momento
def calcular_media_gols(df):
    # Dicionário para guardar o total de gols e jogos de cada time
    stats_times = {}
    
    medias_mandante = []
    medias_visitante = []
    
    for indice, linha in df.iterrows():
        mandante = linha['Mandante']
        visitante = linha['Visitante']
        
        # Pega a média histórica do mandante até agora (se não tiver, assume 1.0)
        m_gols = stats_times.get(mandante, {'gols': 0, 'jogos': 0})
        media_m = m_gols['gols'] / m_gols['jogos'] if m_gols['jogos'] > 0 else 1.0
        medias_mandante.append(media_m)
        
        # Pega a média histórica do visitante até agora
        v_gols = stats_times.get(visitante, {'gols': 0, 'jogos': 0})
        media_v = v_gols['gols'] / v_gols['jogos'] if v_gols['jogos'] > 0 else 1.0
        medias_visitante.append(media_v)
        
        # ATUALIZA os dados DEPOIS de salvar a média (para não usar o dado do jogo atual na previsão dele mesmo!)
        # Isso evita o "Data Leakage" (vazamento de dados)
        stats_times[mandante] = {'gols': m_gols['gols'] + linha['Gols_Mandante'], 'jogos': m_gols['jogos'] + 1}
        stats_times[visitante] = {'gols': v_gols['gols'] + linha['Gols_Visitante'], 'jogos': v_gols['jogos'] + 1}
        
    df['Media_Gols_Mandante'] = medias_mandante
    df['Media_Gols_Visitante'] = medias_visitante
    return df

# Aplica a função
df_inteligente = calcular_media_gols(df_limpo)

print("Agora temos colunas que representam a FORÇA DO ATAQUE antes do jogo começar:")
display(df_inteligente[['Data', 'Mandante', 'Visitante', 'Media_Gols_Mandante', 'Media_Gols_Visitante']].tail())
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Carregar os dados que salvamos anteriormente
df = pd.read_csv('/Users/victorhugopradateixeira/SoccerPredict/dados/brasileirao_dados_processados.csv')

# Garantir que não temos linhas vazias que quebram o modelo
df = df.dropna()

print(f"Dados carregados: {df.shape[0]} partidas prontas para análise.")

# 2. Definir as "Features" (O que o modelo VÊ) e o "Alvo" (O que ele deve ADIVINHAR)
# Por enquanto, vamos usar apenas as médias de gols que criamos
features = ['Media_Gols_Mandante', 'Media_Gols_Visitante']
target = 'Alvo' # 0=Empate, 1=Mandante, 2=Visitante

X = df[features]
y = df[target]

# 3. Separar Treino e Teste (O jeito CERTO para Futebol)
# ATENÇÃO: Em futebol, NÃO podemos embaralhar (shuffle) os dados aleatoriamente.
# Temos que treinar com o PASSADO e testar com o FUTURO.
# Vamos usar os primeiros 80% dos jogos para treinar e os 20% finais para testar.

ponto_corte = int(len(df) * 0.8)

X_treino = X.iloc[:ponto_corte]
y_treino = y.iloc[:ponto_corte]

X_teste = X.iloc[ponto_corte:]
y_teste = y.iloc[ponto_corte:]

print(f"\nTreinando com {len(X_treino)} jogos.")
print(f"Testando com {len(X_teste)} jogos recentes.")

# 4. Criar e Treinar o Modelo (Random Forest)
# Random Forest cria várias "árvores de decisão" e vota no melhor resultado.
modelo = RandomForestClassifier(n_estimators=100, random_state=42)
modelo.fit(X_treino, y_treino)

print("\nModelo treinado com sucesso! 🧠")

# 5. Avaliar o Modelo
previsoes = modelo.predict(X_teste)
acuracia = accuracy_score(y_teste, previsoes)

print(f"\nAcurácia do Modelo: {acuracia:.2%}") # Mostra em porcentagem
print("\nRelatório Detalhado:")
print(classification_report(y_teste, previsoes, target_names=['Empate', 'Mandante', 'Visitante']))

# 6. Teste Real: Vamos simular um jogo
# Imagine um Mandante forte (média 2.0 gols) contra um Visitante fraco (média 0.5 gols)
novo_jogo = pd.DataFrame({
    'Media_Gols_Mandante': [2.0],
    'Media_Gols_Visitante': [0.5]
})

previsao_jogo = modelo.predict(novo_jogo)
probs = modelo.predict_proba(novo_jogo) # Probabilidades

mapa_reverso = {0: 'Empate', 1: 'Vitória Mandante', 2: 'Vitória Visitante'}
print("-" * 30)
print(f"Simulação: Mandante Forte vs Visitante Fraco")
print(f"Previsão da IA: {mapa_reverso[previsao_jogo[0]]}")
print(f"Probabilidades: {probs[0]}")
```
